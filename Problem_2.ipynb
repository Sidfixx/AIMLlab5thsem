```python
from collections import deque

def water_jug_bfs():
    capacity_a = 4
    capacity_b = 3

    start = (0, 0)
    queue = deque([(start, [])])
    visited = {start}

    while queue:
        (a, b), path = queue.popleft()

        # Goal condition
        if a == 2 or b == 2:
            return path + [(a, b)]

        next_states = []

        # Fill Jug A
        next_states.append((capacity_a, b))

        # Fill Jug B
        next_states.append((a, capacity_b))

        # Empty Jug A
        next_states.append((0, b))

        # Empty Jug B
        next_states.append((a, 0))

        # Pour A -> B
        amount = min(a, capacity_b - b)
        next_states.append((a - amount, b + amount))

        # Pour B -> A
        amount = min(b, capacity_a - a)
        next_states.append((a + amount, b - amount))

        for state in next_states:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [(a, b)]))

# Solve
solution = water_jug_bfs()

for i, state in enumerate(solution):
    print(f"Step {i}: {state}")
```
