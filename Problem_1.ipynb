```python
import networkx as nx
import matplotlib.pyplot as plt

# Adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

# Create graph
G = nx.DiGraph(graph)

# BFS and DFS
bfs = list(nx.bfs_tree(G, 'A'))
dfs = list(nx.dfs_tree(G, 'A'))

# Print traversal orders
print("BFS Traversal:", " -> ".join(bfs))
print("DFS Traversal:", " -> ".join(dfs))

# Node positions
pos = {
    'A': (0, 2),
    'B': (-1, 1),
    'C': (1, 1),
    'D': (-1.5, 0),
    'E': (-0.5, 0),
    'F': (1, 0)
}

# ---------------- BFS Diagram ----------------
plt.figure(figsize=(8, 6))

nx.draw(
    G, pos,
    with_labels=False,
    node_color="lightblue",
    node_size=2000,
    edge_color="gray",
    arrows=False
)

bfs_labels = {
    node: f"{node}\n({i})"
    for i, node in enumerate(bfs, 1)
}

nx.draw_networkx_labels(
    G, pos,
    labels=bfs_labels,
    font_size=12,
    font_weight="bold"
)

plt.title("BFS Traversal Order")
plt.axis("off")
plt.show()


# ---------------- DFS Diagram ----------------
plt.figure(figsize=(8, 6))

nx.draw(
    G, pos,
    with_labels=False,
    node_color="lightgreen",
    node_size=2000,
    edge_color="gray",
    arrows=False
)

dfs_labels = {
    node: f"{node}\n({i})"
    for i, node in enumerate(dfs, 1)
}

nx.draw_networkx_labels(
    G, pos,
    labels=dfs_labels,
    font_size=12,
    font_weight="bold"
)

plt.title("DFS Traversal Order")
plt.axis("off")
plt.show()
```
