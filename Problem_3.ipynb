```python


# Graph representing cities and roads
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D'],
    'F': ['G'],
    'G': ['F']
}


# DFS to check whether a route exists between two cities
def route_exists(graph, start, destination):
    visited = set()

    def dfs(city):
        if city == destination:
            return True

        visited.add(city)

        for neighbor in graph[city]:
            if neighbor not in visited:
                if dfs(neighbor):
                    return True

        return False

    return dfs(start)


# DFS to find all cities in one connected component
def find_component(graph, start, visited):
    component = []

    def dfs(city):
        visited.add(city)
        component.append(city)

        for neighbor in graph[city]:
            if neighbor not in visited:
                dfs(neighbor)

    dfs(start)
    return component


# Check route between two cities
start_city = 'A'
destination_city = 'E'

if route_exists(graph, start_city, destination_city):
    print(f"Route exists from {start_city} to {destination_city}")
else:
    print(f"No route exists from {start_city} to {destination_city}")


# Find all connected components
visited = set()
components = []

for city in graph:
    if city not in visited:
        component = find_component(graph, city, visited)
        components.append(component)

print("\nConnected Components:")

for i, component in enumerate(components, 1):
    print(f"Component {i}: {component}")
```
